## Connecting to DB in python

```python
from sqlite3 import connect
import pandas as pd

with connect('data/soccer.sqlite') as conn:
  query = """
    SELECT *
    FROM Player
    LIMIT 10
  """

  result = pd.read_sql(
    query,
    con=conn
  )

result
  # do stuff with that connection
```

## Basic search

```python
name = "Helmut%"

with connect("data/soccer.sqlite") as conn:
    query = f"""
        SELECT *
          FROM Player
         WHERE player_name LIKE '{name}'
    """
    players_df = pd.read_sql(query, con=conn)

players_df.head()
```
