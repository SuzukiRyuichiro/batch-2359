```python
def fetch_population_of_a_country(country, year):
    url = f"https://api.worldbank.org/v2/country/{country}/indicator/SP.POP.TOTL?date={year}&format=json"
    response = requests.get(url)
    data = response.json()

    if response.ok != True or data[1] == None:
        print('Failed')
        return None

    return data[1][0]["value"]
```

```python
fetch_population_of_a_country('JP', 2023)
```

```python
# iterate over each row of the df
# on each row, use the country code to get the population
# add that info as a new column

for index, row in merged_df.head(3).iterrows():
    population = fetch_population_of_a_country(row["Code"], 2012)
    merged_df.loc[index, 'Population'] = population
```

```python
# Create empty lists as placeholders for the codes, names and populations
country_codes = []
country_names = []
populations = []

# Start at page 1 and start a while loop (we don't know how many pages we'll get)
page = 1
while True:
    url = f"https://api.worldbank.org/v2/country/all/indicator/SP.POP.TOTL"

    params = {
        "format": "json",
        "date": year,
        "page": page,  # This is new: we pass the page we want as a param
    }

    response = requests.get(url=url, params=params).json()

    # Add the codes, names and populations to our placeholders
    country_codes.extend([result["country"]["id"] for result in response[1]])
    country_names.extend([result["country"]["value"] for result in response[1]])
    populations.extend(result["value"] for result in response[1])

    # Check if this wasn't the last page
    if page == response[0]["pages"]:
        break
    # Get ready for the next page
    page += 1
```

```python
population_df = pd.DataFrame(
    {"Code": country_codes, "Name": country_names, "Population": populations}
)
population_df.merge(merged_df, left_on='Code', right_on='Code')
```

```python
new_order = ["Name",	"Code",	"Full Name",	"Region",	"Population"]

new_order += list(merged_df.columns[:-5])
new_order
```

```python
merged_df = merged_df[new_order]
```

```python
merged_df.to_csv("data/emissions_enriched.csv", index=False)
```
