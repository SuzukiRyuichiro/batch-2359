```python
def fetch_population_of_a_country(country, year):
    url = f"https://api.worldbank.org/v2/country/{country}/indicator/SP.POP.TOTL?date={year}&format=json"
    response = requests.get(url)
    data = response.json()

    if response.ok != True or data[1] == None:
        print('Failed')
        return None

    return data[1][0]["value"]
```

```python
fetch_population_of_a_country('JP', 2023)
```
